```python
#reflection
import matplotlib.pyplot as plt
import numpy as np

# Original points of a triangle
points = np.array([
    [1, 1],
    [3, 1],
    [2, 3],
    [1, 1]  # Closing the triangle
])

# Reflection functions
def reflect_x(points):
    reflected = points.copy()
    reflected[:, 1] = -reflected[:, 1]
    return reflected

def reflect_y(points):
    reflected = points.copy()
    reflected[:, 0] = -reflected[:, 0]
    return reflected

def reflect_origin(points):
    return -points

def reflect_y_eq_x(points):
    reflected = points.copy()
    reflected[:, 0], reflected[:, 1] = points[:, 1], points[:, 0]
    return reflected

# Apply reflections
reflected_x = reflect_x(points)
reflected_y = reflect_y(points)
reflected_origin = reflect_origin(points)
reflected_y_eq_x = reflect_y_eq_x(points)

# Plotting
plt.plot(points[:, 0], points[:, 1], 'b-', label='Original')
plt.plot(reflected_x[:, 0], reflected_x[:, 1], 'r--', label='Reflection about X-axis')
plt.plot(reflected_y[:, 0], reflected_y[:, 1], 'g--', label='Reflection about Y-axis')
plt.plot(reflected_origin[:, 0], reflected_origin[:, 1], 'm--', label='Reflection about Origin')
plt.plot(reflected_y_eq_x[:, 0], reflected_y_eq_x[:, 1], 'c--', label='Reflection about y=x')

plt.legend()
plt.title('2D Reflection of a Triangle')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
#rotation

import matplotlib.pyplot as plt
import numpy as np

# Original points of a triangle
points = np.array([
    [1, 1],
    [3, 1],
    [2, 3],
    [1, 1]  # Closing the triangle
])

# Rotation angle in degrees
angle_deg = 45
angle_rad = np.deg2rad(angle_deg)  # convert to radians

# Function to rotate points
def rotate(points, angle):
    rotation_matrix = np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle),  np.cos(angle)]
    ])
    return points.dot(rotation_matrix.T)  # matrix multiplication

# Rotate the original points
rotated_points = rotate(points, angle_rad)

# Plot original triangle
plt.plot(points[:, 0], points[:, 1], 'b-', label='Original')

# Plot rotated triangle
plt.plot(rotated_points[:, 0], rotated_points[:, 1], 'm--', label=f'Rotated {angle_deg}°')

plt.legend()
plt.title('2D Rotation of a Triangle')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
#scalling

import matplotlib.pyplot as plt
import numpy as np

# Original points of a triangle
points = np.array([
    [1, 1],
    [3, 1],
    [2, 3],
    [1, 1]  # Closing the triangle
])

# Scaling factors
sx, sy = 2, 0.5

# Function to scale points
def scale(points, sx, sy):
    scaling_matrix = np.array([sx, sy])
    return points * scaling_matrix

# Scale the original points
scaled_points = scale(points, sx, sy)

# Plot original triangle
plt.plot(points[:, 0], points[:, 1], 'b-', label='Original')

# Plot scaled triangle
plt.plot(scaled_points[:, 0], scaled_points[:, 1], 'g--', label='Scaled')

plt.legend()
plt.title('2D Scaling of a Triangle')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
#shearing

import matplotlib.pyplot as plt
import numpy as np

# Original points of a triangle
points = np.array([
    [1, 1],
    [3, 1],
    [2, 3],
    [1, 1]  # Closing the triangle
])

# Shear factors
shx = 1.0  # shear along x-axis
shy = 0.5  # shear along y-axis

# Function to shear points along x
def shear_x(points, shx):
    # x' = x + shx * y
    x_new = points[:, 0] + shx * points[:, 1]
    y_new = points[:, 1]
    return np.column_stack((x_new, y_new))

# Function to shear points along y
def shear_y(points, shy):
    # y' = y + shy * x
    x_new = points[:, 0]
    y_new = points[:, 1] + shy * points[:, 0]
    return np.column_stack((x_new, y_new))

# Shear the points (you can try either shear_x or shear_y)
sheared_points_x = shear_x(points, shx)
sheared_points_y = shear_y(points, shy)

# Plot original triangle
plt.plot(points[:, 0], points[:, 1], 'b-', label='Original')

# Plot sheared triangles
plt.plot(sheared_points_x[:, 0], sheared_points_x[:, 1], 'r--', label='Sheared X')
plt.plot(sheared_points_y[:, 0], sheared_points_y[:, 1], 'g--', label='Sheared Y')

plt.legend()
plt.title('2D Shearing of a Triangle')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
#translation

import matplotlib.pyplot as plt
import numpy as np

# Original points of a triangle
points = np.array([
    [1, 1],
    [3, 1],
    [2, 3],
    [1, 1]  # Closing the triangle
])

# Translation vector
tx, ty = 2, 3

# Function to translate points
def translate(points, tx, ty):
    translation_matrix = np.array([tx, ty])
    return points + translation_matrix

# Translate the original points
translated_points = translate(points, tx, ty)

# Plot original triangle
plt.plot(points[:, 0], points[:, 1], 'b-', label='Original')

# Plot translated triangle
plt.plot(translated_points[:, 0], translated_points[:, 1], 'r--', label='Translated')

plt.legend()
plt.title('2D Translation of a Triangle')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```
